Problem - 1: Array Creation:
Complete the following Tasks:
1. Initialize an empty array with size 2X2.
2. Initialize an all one array with size 4X2.
3. Return a new array of given shape and type, filled with fill value.{Hint: np.full}
4. Return a new array of zeros with same shape and type as a given array.{Hint: np.zeros like}
5. Return a new array of ones with same shape and type as a given array.{Hint: np.ones like}
6. For an existing list new_list = [1,2,3,4] convert to an numpy array.{Hint: np.array()}

```python
import numpy as np

arr1 = np.empty((2, 2))
print("1. Empty Array:\n", arr1)

arr2 = np.ones((4, 2))
print("\n2. All One Array:\n", arr2)

arr3 = np.full((3, 3), 5, dtype=int)
print("\n3. Full Array:\n", arr3)

original_array = np.array([[1, 2], [3, 4]])
arr4 = np.zeros_like(original_array)
print("\n4. Zeros Like Array:\n", arr4)

arr5 = np.ones_like(original_array)
print("\n5. Ones Like Array:\n", arr5)

new_list = [1, 2, 3, 4]
arr6 = np.array(new_list)
print("\n6. Converted NumPy Array:\n", arr6)
```

{Hint:try np.concatenate() or np.vstack() functions.

```python
import numpy as np

x = np.array([[1, 2], [3, 5]])
y = np.array([[5, 6], [7, 8]])
v = np.array([9, 10])
w = np.array([11, 12])

add = np.add(x, y)
print("1. Addition:\n", add)

subtract = np.subtract(x, y)
print("\n2. Subtraction:\n", subtract)

multiply = x * 5
print("\n3. Multiplication:\n", multiply)

square = np.square(x)
print("\n4. Square of Each Element:\n", square)

dot_vw = np.dot(v, w)
dot_xv = np.dot(x, v)
dot_xy = np.dot(x, y)

print("\n5. Dot Product of v and w:\n", dot_vw)
print("Dot Product of x and v:\n", dot_xv)
print("Dot Product of x and y:\n", dot_xy)

concat_xy = np.concatenate((x, y), axis=0)
print("\n6. Concatenation of x and y (Rows):\n", concat_xy)

concat_vw = np.vstack((v, w))
print("\nConcatenation of v and w (Rows):\n", concat_vw)

try:
    concat_xv = np.concatenate((x, v))
    print("\n7. Concatenation of x and v:\n", concat_xv)
except ValueError as e:
    print("\n7. Error while concatenating x and v:")
    print(e)
```

• Solve the following system of Linear equation using Inverse Methods.

2x − 3y + z = −1
x − y + 2z = −3
3x + y − z = 9

{Hint: First use Numpy array to represent the equation in Matrix form. Then Solve for: AX = B}

```python

import numpy as np

A = np.array([[3, 4], [7, 8]])
B = np.array([[5, 3], [2, 1]])

A_inv = np.linalg.inv(A)

result1 = np.dot(A, A_inv)

print("1. A * A inverse:\n", np.round(result1).astype(int))

print("Identity Matrix:\n", np.eye(2, dtype=int))

print("Is A * A inverse equal to I?",
      np.allclose(result1, np.eye(2)))


AB = np.dot(A, B)
BA = np.dot(B, A)

print("\n2. AB:\n", AB)
print("BA:\n", BA)

print("Is AB not equal to BA?",
      not np.array_equal(AB, BA))


left_side = np.transpose(AB)

right_side = np.dot(np.transpose(B), np.transpose(A))

print("\n3. (AB) Transpose:\n", left_side)
print("B Transpose * A Transpose:\n", right_side)

print("Are both sides equal?",
      np.allclose(left_side, right_side))



C = np.array([
    [2, -3, 1],
    [1, -1, 2],
    [3, 1, -1]
])

D = np.array([-1, -3, 9])

C_inv = np.linalg.inv(C)

X = np.dot(C_inv, D)

print("\n4. Solution Using np.linalg.inv():")

print("x =", X[0])
print("y =", X[1])
print("z =", X[2])


X2 = np.linalg.solve(C, D)

print("\n5. Solution Using np.linalg.solve():")

print("x =", X2[0])
print("y =", X2[1])
print("z =", X2[2])
```
